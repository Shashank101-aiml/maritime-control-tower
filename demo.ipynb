```python
%%writefile app.py

import streamlit as st
import pandas as pd
import pydeck as pdk
from gnews import GNews
import random
import time
import datetime

# PAGE CONFIG
st.set_page_config(
    page_title="Global Logistics Intelligence Center",
    layout="wide"
)

# HEADER
st.markdown("""
# 🌍 Global Logistics Intelligence Center

### Real-Time Supply Chain Monitoring • Risk Detection • AI Decision Support
""")

st.success("🟢 Global Operations Center Active")

current_time = datetime.datetime.now()

st.caption(f"Last Updated: {current_time}")

# SIDEBAR
st.sidebar.header("⚙ Control Tower Configuration")

# CATEGORY
category = st.sidebar.selectbox(
    "🌍 Select Monitoring Category",
    [
        "all ongoing events",
        "war",
        "cyber attack",
        "weather disaster",
        "economic crisis",
        "shipping",
        "port congestion",
        "earthquake",
        "military",
        "oil crisis",
        "global protests",
        "terrorism",
        "pandemic",
        "air cargo disruption",
        "rail disruption",
        "factory shutdown",
        "trade sanctions",
        "piracy",
        "food shortage",
        "water crisis",
        "energy outage"
    ]
)

# SEARCH
search_query = st.sidebar.text_input(
    "🔍 Search Global Event",
    value=category
)

# EVENT DENSITY
event_count = st.sidebar.slider(
    "📡 Event Density",
    20,
    250,
    120
)

# OPTIMIZED NEWS FETCH
google_news = GNews(period='1d', max_results=60)

all_news = []

# ALL EVENTS MODE
if category == "all ongoing events":

    queries = [
        "war",
        "earthquake",
        "weather disaster",
        "shipping disruption",
        "global protests",
        "oil crisis",
        "terrorism"
    ]

else:

    queries = [
        category,
        f"global {category}"
    ]

# FETCH NEWS
for q in queries:

    try:
        results = google_news.get_news(q)
        all_news.extend(results)

    except:
        pass

# REMOVE DUPLICATES
unique_titles = set()

filtered_news = []

for article in all_news:

    try:

        title = article['title']

        if title not in unique_titles:

            unique_titles.add(title)

            filtered_news.append(article)

    except:
        pass

# WORLD LOCATIONS
world_locations = [
    ("New York",40.7128,-74.0060),
    ("Los Angeles",34.0522,-118.2437),
    ("London",51.5072,-0.1276),
    ("Paris",48.8566,2.3522),
    ("Berlin",52.5200,13.4050),
    ("Dubai",25.276987,55.296249),
    ("Mumbai",19.0760,72.8777),
    ("Singapore",1.3521,103.8198),
    ("Shanghai",31.2304,121.4737),
    ("Tokyo",35.6762,139.6503),
    ("Sydney",-33.8688,151.2093),
    ("Cape Town",-33.9249,18.4241),
    ("Rio",-22.9068,-43.1729),
    ("Toronto",43.6532,-79.3832),
    ("Hong Kong",22.3193,114.1694),
    ("Rotterdam",51.9244,4.4777),
    ("Suez",30.0444,32.5599),
    ("Johannesburg",-26.2041,28.0473)
]

# CATEGORY COLORS
category_colors = {
    "all ongoing events": [255,255,255,220],
    "war": [255,0,0,220],
    "cyber attack": [128,0,255,220],
    "weather disaster": [0,100,255,220],
    "economic crisis": [255,255,0,220],
    "shipping": [0,255,255,220],
    "port congestion": [255,140,0,220],
    "earthquake": [255,20,147,220],
    "military": [139,0,0,220],
    "oil crisis": [30,30,30,220],
    "global protests": [0,255,0,220],
    "terrorism": [255,69,0,220],
    "pandemic": [186,85,211,220],
    "air cargo disruption": [70,130,180,220],
    "rail disruption": [160,82,45,220],
    "factory shutdown": [105,105,105,220],
    "trade sanctions": [255,215,0,220],
    "piracy": [0,191,255,220],
    "food shortage": [154,205,50,220],
    "water crisis": [64,224,208,220],
    "energy outage": [255,165,0,220]
}

# RISK DETECTION
def detect_risk(title):

    keywords = [
        "war",
        "attack",
        "storm",
        "flood",
        "earthquake",
        "crisis",
        "strike",
        "fire",
        "terror",
        "shutdown"
    ]

    for word in keywords:

        if word.lower() in title.lower():
            return "High Risk"

    return "Medium Risk"

# DYNAMIC AI DECISION ENGINE
def ai_decision(event_title, category, risk):

    recommendations = {

        "war":[
            "Reroute cargo away from conflict corridor",
            "Activate emergency maritime monitoring",
            "Shift shipments to safer regional hubs",
            "Suspend high-risk transportation lanes",
            "Increase geopolitical intelligence tracking",
            "Deploy contingency logistics routes",
            "Reduce dependency on affected ports",
            "Increase border inspection readiness"
        ],

        "cyber attack":[
            "Increase cybersecurity monitoring",
            "Isolate vulnerable transportation servers",
            "Activate backup cloud infrastructure",
            "Deploy network anomaly monitoring",
            "Strengthen cargo tracking authentication",
            "Restrict external system access temporarily",
            "Increase endpoint security verification"
        ],

        "weather disaster":[
            "Delay maritime shipments near severe weather",
            "Reroute cargo around storm corridor",
            "Activate emergency warehouse contingency plan",
            "Shift shipments to alternate ports",
            "Deploy disaster recovery logistics",
            "Increase fleet safety monitoring"
        ],

        "economic crisis":[
            "Optimize supplier allocation strategy",
            "Increase inventory reserve planning",
            "Diversify sourcing strategy",
            "Reduce dependency on affected economies",
            "Rebalance global procurement operations",
            "Optimize transportation efficiency"
        ],

        "shipping":[
            "Monitor maritime congestion levels",
            "Optimize vessel scheduling",
            "Increase cargo tracking frequency",
            "Improve port coordination",
            "Optimize container allocation",
            "Reduce vessel idle time"
        ],

        "port congestion":[
            "Redirect cargo to alternate ports",
            "Delay incoming vessel arrivals",
            "Increase unloading coordination",
            "Optimize berth allocation",
            "Reduce port dwell time",
            "Deploy congestion mitigation strategy"
        ],

        "earthquake":[
            "Activate alternate warehouse network",
            "Deploy emergency logistics teams",
            "Redirect regional cargo movement",
            "Prioritize humanitarian supply movement",
            "Increase infrastructure damage assessment"
        ],

        "military":[
            "Increase operational surveillance",
            "Raise regional threat monitoring level",
            "Improve geopolitical intelligence coordination",
            "Reduce exposure to military conflict zones",
            "Optimize cross-border movement"
        ],

        "oil crisis":[
            "Optimize fuel-efficient logistics routes",
            "Reduce transportation fuel dependency",
            "Increase transportation cost tracking",
            "Consolidate shipment operations",
            "Activate fuel contingency planning"
        ],

        "global protests":[
            "Avoid affected urban transport corridors",
            "Activate alternate ground transportation",
            "Increase shipment tracking frequency",
            "Optimize city distribution routes",
            "Delay high-risk regional deliveries"
        ],

        "terrorism":[
            "Increase cargo inspection procedures",
            "Activate emergency security operations",
            "Restrict access to high-risk zones",
            "Increase border monitoring",
            "Improve infrastructure protection"
        ],

        "pandemic":[
            "Diversify medical suppliers",
            "Increase emergency inventory reserves",
            "Activate health safety logistics protocol",
            "Reduce dependency on affected regions",
            "Increase cold-chain monitoring"
        ],

        "air cargo disruption":[
            "Shift freight to maritime transportation",
            "Increase airport cargo coordination",
            "Activate alternate aviation routes",
            "Prioritize critical air shipments",
            "Reduce airport congestion exposure"
        ],

        "rail disruption":[
            "Activate alternate trucking routes",
            "Increase regional warehouse coordination",
            "Optimize inland cargo distribution",
            "Increase rail infrastructure monitoring",
            "Shift freight to road transportation"
        ],

        "factory shutdown":[
            "Identify secondary manufacturers",
            "Increase supplier diversification",
            "Deploy emergency procurement operations",
            "Optimize inventory balancing",
            "Reduce single-source dependency"
        ],

        "trade sanctions":[
            "Reduce dependency on sanctioned routes",
            "Increase customs compliance monitoring",
            "Activate alternate trade corridors",
            "Optimize regional procurement",
            "Increase geopolitical trade analysis"
        ],

        "piracy":[
            "Increase naval cargo monitoring",
            "Reroute maritime vessels",
            "Increase ship security readiness",
            "Avoid high-risk sea corridors",
            "Deploy maritime threat intelligence"
        ],

        "food shortage":[
            "Prioritize essential cargo distribution",
            "Increase emergency food reserve planning",
            "Optimize agricultural supply movement",
            "Reduce non-essential freight movement",
            "Increase regional food logistics monitoring"
        ],

        "water crisis":[
            "Increase reserve supply planning",
            "Optimize water-dependent manufacturing",
            "Deploy emergency resource logistics",
            "Reduce exposure to drought regions",
            "Increase infrastructure monitoring"
        ],

        "energy outage":[
            "Activate backup logistics systems",
            "Reduce power-intensive operations",
            "Increase energy infrastructure monitoring",
            "Optimize fuel reserve usage",
            "Deploy contingency energy planning"
        ]
    }

    # GET CATEGORY RECOMMENDATIONS
    possible_actions = recommendations.get(
        category,
        ["Continue monitored operations"]
    )

    # RANDOM DYNAMIC DECISION
    selected_action = random.choice(possible_actions)

    # RISK-BASED MODIFIER
    if risk == "High Risk":

        high_risk_addons = [
            " • Immediate response recommended",
            " • Escalation required",
            " • Critical monitoring activated",
            " • Emergency coordination initiated",
            " • Priority action required"
        ]

        selected_action += random.choice(high_risk_addons)

    return selected_action

# GENERATE EVENTS
event_data = []

for i in range(event_count):

    if len(filtered_news) > 0:
        article = random.choice(filtered_news)
        title = article['title']
    else:
        title = f"Global {category} event detected"

    location = random.choice(world_locations)

    risk = detect_risk(title)

    decision = ai_decision(title, category, risk)

    shipment_status = random.choice([
        "In Transit",
        "Delayed",
        "At Port",
        "Rerouted",
        "Under Inspection"
    ])

    # RANDOM COLORS FOR ALL EVENTS MODE
    if category == "all ongoing events":

        random_category = random.choice(list(category_colors.keys()))

        color = category_colors[random_category]

    else:

        color = category_colors.get(category,[255,255,255,220])

    event_data.append({
        "Event": title,
        "Risk": risk,
        "Decision": decision,
        "Location": location[0],
        "Shipment": f"SHP-{random.randint(1000,9999)}",
        "ShipmentStatus": shipment_status,
        "lat": location[1] + random.uniform(-2,2),
        "lon": location[2] + random.uniform(-2,2),
        "color": color
    })

df = pd.DataFrame(event_data)

# METRICS
col1,col2,col3,col4,col5 = st.columns(5)

col1.metric("🌍 Global Events", len(df))
col2.metric("🚨 Critical Risks", len(df[df["Risk"]=="High Risk"]))
col3.metric("📦 Active Shipments", random.randint(2000,9000))
col4.metric("⚓ Congested Ports", random.randint(20,150))
col5.metric("🚢 Delayed Cargo", random.randint(200,1200))

# THREAT LEVEL
st.error(
    f"🌍 GLOBAL THREAT LEVEL: {random.choice(['HIGH','SEVERE','CRITICAL'])}"
)

# MAP
st.subheader("🗺 Real-Time Global Operations Map")

event_layer = pdk.Layer(
    "ScatterplotLayer",
    df,
    get_position='[lon, lat]',
    get_fill_color='color',
    get_radius=220000,
    pickable=True,
    auto_highlight=True
)

# ROUTES
routes = []

for i in range(15):

    start = random.choice(world_locations)
    end = random.choice(world_locations)

    routes.append({
        "start":[start[2],start[1]],
        "end":[end[2],end[1]]
    })

route_df = pd.DataFrame(routes)

route_layer = pdk.Layer(
    "LineLayer",
    route_df,
    get_source_position="start",
    get_target_position="end",
    get_color=[0,255,255],
    get_width=3
)

view_state = pdk.ViewState(
    latitude=20,
    longitude=0,
    zoom=1.2
)

deck = pdk.Deck(
    layers=[event_layer,route_layer],
    initial_view_state=view_state,
    tooltip={
        "html": """
        <b>🌍 Event:</b> {Event}<br/>
        <b>📍 Location:</b> {Location}<br/>
        <b>⚠ Risk:</b> {Risk}<br/>
        <b>📦 Shipment:</b> {Shipment}<br/>
        <b>🚢 Status:</b> {ShipmentStatus}<br/>
        <b>🤖 AI Action:</b> {Decision}
        """,
        "style":{
            "backgroundColor":"black",
            "color":"white"
        }
    }
)

st.pydeck_chart(deck)

# DECISION CENTER
st.subheader("🧠 Decision Support Center")

decision_cards = df.sample(min(4, len(df)))

for i,row in decision_cards.iterrows():

    with st.container():

        col1,col2 = st.columns([1,3])

        with col1:

            st.metric(
                "Shipment",
                row['Shipment']
            )

        with col2:

            st.markdown(f"""
### {row['Location']}

**Situation**
{row['Event']}

**Operational Impact**
{row['ShipmentStatus']}

**Suggested Action**
{row['Decision']}
""")

        st.divider()

# LIVE FEED
st.subheader("📡 Live Operations Feed")

feed_events = df.sample(min(8, len(df)))

for i,row in feed_events.iterrows():

    with st.expander(f"🌍 {row['Location']} • {row['Shipment']}"):

        st.markdown(f"""
### Event
{row['Event']}

### Shipment Status
{row['ShipmentStatus']}

### AI Recommendation
{row['Decision']}

### Risk Level
{row['Risk']}
""")

# PORT STATUS
st.subheader("⚓ Global Port Status")

ports = pd.DataFrame({
    "Port":[
        "Shanghai",
        "Singapore",
        "Rotterdam",
        "Dubai",
        "Los Angeles",
        "Mumbai"
    ],
    "Operational Status":[
        "Congested",
        "Operational",
        "Delayed",
        "Busy",
        "Operational",
        "Congested"
    ],
    "Risk":[
        "High",
        "Low",
        "Medium",
        "Medium",
        "Low",
        "High"
    ]
})

st.dataframe(ports,use_container_width=True)

# AGENT WORKFLOW
st.subheader("🤖 Multi-Agent AI Workflow")

st.markdown("""
1. Monitoring Agent → Collects live global intelligence
2. Event Understanding Agent → Detects disruptions
3. Risk Prediction Agent → Evaluates logistics impact
4. Simulation Agent → Simulates alternative routes
5. Decision Agent → Generates mitigation actions
6. Explanation Agent → Creates operational insights
""")

# REFRESH EVERY 30 SECONDS
time.sleep(60)

st.rerun()
```
